```python
# Import neccesary libraries:
import pandas as pd
import numpy as np

# Download Data:
from sklearn.datasets import fetch_california_housing

# Load the Housing Data Set:
housing = fetch_california_housing()

# Creating a Pandas DataFrame for the features and a series for the target variable:
x = pd.DataFrame(housing.data, columns=housing.feature_names)
y = pd.Series(housing.target, name='med_house_value')

# Displaying the first five rows:
print("First 5 rows of data:")
x.head()
```

```python
# Printing the Feature Names and Checking for missing values:
print("Feature Names:")
print(housing.feature_names)
print("\nMissing Values:")
print(x.isnull().sum())
```

```python
# Summary Statistics:
x.describe()
```

```python
# Splitting the data set into training and test sets:
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Training a linear regression model on unscaled data:
model = LinearRegression()
model.fit(x_train, y_train)

# Making predictions on the test set:
y_pred = model.predict(x_test)

```

```python
# Evaluating model performance using MSE, RMSE, and R^2:
from sklearn.metrics import mean_squared_error, root_mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
rmse = root_mean_squared_error(y_test, y_pred)

r2 = r2_score(y_test, y_pred)

print("Unscaled Data Model:")
print(f"Mean Squared Error: {mse:.2f}")
print(f"Root Squared Error: {rmse:.2f}")
print(f"R² Score: {r2:.2f}")
```

```python
# 8: The R squared score inicates the proportion of the variance in the target variable explained by the model.
# An R squared score close to 1 suggests that there is a very good fit, while an R squared score close to 0 
# indicates that the model failed to capture much variance. My R squared score of 0.58 means the model roughly
# explains 58% of the variation in housing prices. My model captures important iformation but definently leaves
# room for improvement. In class we learned that it is often misleading to compare features as they have different 
# units. In my model, I think that MedInc has the strongest impact as it has the largest possible coefficient. 
# I also noticed that Latitude and and Longitude have strong effects and that higher income usually correlates with
# higher house value. Certain geographical locations also significantly influence the price. The RMSE tells me that 
# the predictions are off by about 0.75 (in units of $100,000). This means that the average prediction error is about
# $75,000.

```

```python
# Selecting Three Features:
selected_features = ["MedInc", "AveRooms", "Latitude"]
x_selected = x[selected_features]

# I selected MenInc, AveRooms, and Latitude because median income strongly influences affordability, average rooms
# reflect house size, and Latitude explains the price changes in different locations.

# Training a new linear regression model:
x_train_s, x_test_s, y_train_s, y_test_s = train_test_split(x_selected, y, test_size=0.2, random_state=42)

simple_model = LinearRegression()
simple_model.fit(x_train_s, y_train_s)

y_pred_s = simple_model.predict(x_test_s)

# Evaluating the performance of the simplified model:
mse_s = mean_squared_error(y_test_s, y_pred_s)
rmse_s = np.sqrt(mse_s)
r2_s = r2_score(y_test_s, y_pred_s)

mse_s, rmse_s, r2_s

print("Simplified Data Model:")
print(f"Mean Squared Error: {mse_s:.2f}")
print(f"Root Squared Error: {rmse_s:.2f}")
print(f"R² Score: {r2_s:.2f}")

# So in this model, the RMSE is slightly lower than the R squared score in the first model.

# 16: The simplified model performs slightly worse than the full model. The full model has an R squared score of 0.58
# which means the model roughly explains 58% of the variation. The simplified model has a R squared score of 0.47 which 
# means the model roughly explains only 47% of the variation. The full model captures more information and produces better 
# predictions while the simplified model is easier to interpret and build but is less accurate. The specific situation would dictate which 
# model I would prefer to use. For example, if accuracy was the most important factor, I would choose to use the full model 
# rather than the simplified model.



```
